# Exploratory Data Analysis (EDA)

This notebook performs exploratory data analysis on the NorthBay Living sales data.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Set style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
```

## Load Data

```python
# Load processed data
df = pd.read_csv('../data/processed/processed_data.csv')
df['Date'] = pd.to_datetime(df['Date'])

print(f"Dataset shape: {df.shape}")
print(f"\nDate range: {df['Date'].min()} to {df['Date'].max()}")
print(f"\nNumber of SKUs: {df['SKU'].nunique()}")
print(f"Number of Categories: {df['Category'].nunique()}")
```

## Data Overview

```python
# Display first few rows
df.head()
```

```python
# Data types and missing values
df.info()
```

```python
# Summary statistics
df.describe()
```

## Sales Analysis

```python
# Total sales over time
daily_sales = df.groupby('Date')['Units Sold'].sum().reset_index()

fig = px.line(daily_sales, x='Date', y='Units Sold', title='Total Daily Sales')
fig.show()
```

```python
# Monthly sales trend
df['Month_Year'] = df['Date'].dt.to_period('M')
monthly_sales = df.groupby('Month_Year')['Units Sold'].sum().reset_index()
monthly_sales['Month_Year'] = monthly_sales['Month_Year'].astype(str)

fig = px.bar(monthly_sales, x='Month_Year', y='Units Sold', title='Monthly Sales')
fig.show()
```

## Top Selling Products

```python
# Top 10 SKUs by total sales
sku_sales = df.groupby('SKU')['Units Sold'].sum().sort_values(ascending=False).head(10)

fig = px.bar(x=sku_sales.index, y=sku_sales.values, 
             title='Top 10 SKUs by Total Sales',
             labels={'x': 'SKU', 'y': 'Total Units Sold'})
fig.show()
```

```python
# Bottom 10 SKUs by total sales
sku_sales_bottom = df.groupby('SKU')['Units Sold'].sum().sort_values(ascending=True).head(10)

fig = px.bar(x=sku_sales_bottom.index, y=sku_sales_bottom.values,
             title='Bottom 10 SKUs by Total Sales',
             labels={'x': 'SKU', 'y': 'Total Units Sold'})
fig.show()
```

## Category Analysis

```python
# Sales by category
category_sales = df.groupby('Category')['Units Sold'].sum().sort_values(ascending=False)

fig = px.pie(values=category_sales.values, names=category_sales.index,
             title='Sales Distribution by Category')
fig.show()
```

```python
# Category-wise monthly sales
category_monthly = df.groupby(['Category', 'Month_Year'])['Units Sold'].sum().reset_index()
category_monthly['Month_Year'] = category_monthly['Month_Year'].astype(str)

fig = px.line(category_monthly, x='Month_Year', y='Units Sold', color='Category',
              title='Category-wise Monthly Sales Trend')
fig.show()
```

## Seasonal Trends

```python
# Sales by season
season_sales = df.groupby('Season')['Units Sold'].sum().sort_values(ascending=False)

fig = px.bar(x=season_sales.index, y=season_sales.values,
             title='Sales by Season',
             labels={'x': 'Season', 'y': 'Total Units Sold'})
fig.show()
```

```python
# Sales by day of week
df['Day_Name'] = df['Date'].dt.day_name()
dow_sales = df.groupby('Day_Name')['Units Sold'].sum().reindex(
    ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
)

fig = px.bar(x=dow_sales.index, y=dow_sales.values,
             title='Sales by Day of Week',
             labels={'x': 'Day', 'y': 'Total Units Sold'})
fig.show()
```

## Promotion Impact

```python
# Sales with vs without promotion
promo_sales = df.groupby('Promotion')['Units Sold'].mean()

fig = px.bar(x=promo_sales.index, y=promo_sales.values,
             title='Average Sales: Promotion vs Non-Promotion',
             labels={'x': 'Promotion', 'y': 'Average Units Sold'})
fig.show()
```

## Holiday Impact

```python
# Sales on holidays vs non-holidays
holiday_sales = df.groupby('Holiday')['Units Sold'].mean()

fig = px.bar(x=holiday_sales.index, y=holiday_sales.values,
             title='Average Sales: Holiday vs Non-Holiday',
             labels={'x': 'Holiday', 'y': 'Average Units Sold'})
fig.show()
```

## Correlation Analysis

```python
# Correlation heatmap
numeric_cols = ['Units Sold', 'Revenue', 'Price', 'Cost', 'Promotion', 'Holiday']
correlation_matrix = df[numeric_cols].corr()

fig = px.imshow(correlation_matrix, 
                text_auto=True,
                title='Correlation Matrix',
                color_continuous_scale='RdBu')
fig.show()
```

## Price Analysis

```python
# Price distribution
fig = px.histogram(df, x='Price', nbins=50, title='Price Distribution')
fig.show()
```

```python
# Price vs Units Sold scatter
fig = px.scatter(df.sample(min(10000, len(df))), x='Price', y='Units Sold',
                 title='Price vs Units Sold',
                 trendline='ols')
fig.show()
```
